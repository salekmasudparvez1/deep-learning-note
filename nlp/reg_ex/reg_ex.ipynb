````python
#  \ => Escape character. It allows you to search for special characters like $ or . or + etc. it's mean => "\the_thing_i_want_to_escape"



# 1. \d  -> digits ->[Note:\d finds each digit separately.\d+ groups consecutive digits into a single match.]
# 2. \w  -> word chars (letters, digits, _)->[Note:\w finds each word separately.\w+ groups consecutive words into a single match.]
# 3. \s  -> whitespace [Note:\s finds each whitespace separately.\s+ groups consecutive whitespaces into a single match.]
# 4. \D \W \S -> NOT digit / word / space
# 5. \. \$ \+ \* \? \( \) \[ \] \{ \} \| \^ \\ -> literal special chars
# 6. \b  -> word boundary
# 7. \n \t \r -> newline, tab, carriage return (strings)
# 8. \. in ML file paths / regex for extensions like .csv, .pkl
# 9. \. in tokenizers to escape regex metachars
# 10. \ in Python strings for special chars and path escaping

txt = "Hi This is Salek Masud Parvez, My age is 23 in 2026. Price: $50. Path: C:\\Users\\AI"

# print(".      :", re.findall(r"\.", txt))
print("digits      :", re.findall(r"\d+", txt))
print("words       :", re.findall(r"\w+", txt)[:5])
print("spaces      :", len(re.findall(r"\s", txt)))
print("non-digits  :", re.findall(r"\D+", txt)[:2])
print("literal $   :", re.findall(r"\$", txt))
print("literal .   :", re.findall(r"\.", txt))
print("literal \\  :", re.findall(r"\\", txt))
print("boundary    :", re.findall(r"\bAI\b", txt))
print("extension   :", re.findall(r"\.\w+", "model.pkl data.csv"))
print("newline/tab :", repr("a\nb\tc"))

# if we want to have custom number of digit or word or white space
print("custom_digits      :", re.findall(r"\d{4}", txt)) # Exactly the specified number of occurrences
````

**. Any character (except newline character)**

- `.` = **any single character except `\n`**
- `\.` = **literal dot only** (used for file extensions, decimals)
- `re.DOTALL` flag makes `.` also match newlines
- Common in ML for **tokenization, cleaning, pattern extraction, and tag removal**


````python

txt = "cat bat rat\nmat hat"

# 1. Basic: . matches any single char (except \n)
print(re.findall(r"c.t", txt))          # ['cat']

# 2. . as wildcard between letters
print(re.findall(r".at", txt))          # ['cat','bat','rat','mat','hat'] (not across \n)

# 3. .* non-greedy
print(re.findall(r"c.*?t", txt))        # ['cat']

# 4. .+ greedy: match everything on a line
print(re.findall(r".+", txt))           # ['cat bat rat', 'mat hat']

# 5. . with re.DOTALL -> matches newline too
print(re.findall(r".+", txt, re.DOTALL))# ['cat bat rat\nmat hat']

# 6. Escaped \. matches ONLY literal dot (opposite of .)
print(re.findall(r"\.", "a.b.c"))       # ['.', '.']
print(re.findall(r".", "a.b.c"))        # ['a','.','b','.','c']

# 7. ML use: match any token/char in a sequence
print(re.findall(r"^...$", "cat"))      # ['cat']  (exactly 3 chars)

# 8. ML use(<.*?>): remove any char after a marker (e.g., tags-><tag>,<div>,<text>... etc)[Greedy: <.*>Matches from the first < to the last >Lazy: <.*?> Matches each tag separately, stopping at the nearest possible >]

print(re.sub(r"<.*>", "", "hello <tag>world</tag>"))  # 'hello '
print(re.sub(r"<.*?>", "", "hello <tag>world</tag>"))  # 'hello world'
````

### Asterics

In Regex, asterisk (`*`) means “match zero or more repetitions of the preceding character or pattern.”

## How does `*` work?

| Regex | Meaning                                              | Matches                    |
| ----- | ---------------------------------------------------- | -------------------------- |
| `a*`  | Zero or more `a` characters                          | `""`, `a`, `aa`, `aaa`     |
| `\d*` | Zero or more digits                                  | `""`, `1`, `123`           |
| `\w*` | Zero or more word characters                         | `""`, `abc`, `user123`     |
| `.*`  | Zero or more characters (except newlines by default) | `""`, `hello`, `hello123!` |

- `a*` → zero or more `a` characters.
- `a+` → one or more `a` characters.

## Greedy behavior of `*`

By default, `*` is greedy: it tries to match as much as possible.

```
print(re.findall(r"<.*>", "<h1>Hello</h1>")) => ['<h1>Hello</h1>']
```

Here, `.*` consumes as much text as it can while allowing the full pattern to match. Adding `?` makes it lazy, so it matches less text.

## Remember these three symbols

| Symbol | Meaning                                 | Example |
| ------ | --------------------------------------- | ------- |
| `*`    | Zero or more                            | `a*`    |
| `+`    | One or more                             | `a+`    |
| `?`    | Zero or one (when used as a quantifier) | `a?`    |

Easy trick: `*` allows nothing, one occurrence, or many occurrences.


````python
# Practice 1: Identify email-like text using .*@.*\..*
txt = "akash@gmail.com"

x = re.findall(r".*@.*\..*", txt)
print(x)
# Output: ['akash@gmail.com']

# Breakdown:
# .*  => Match zero or more characters before @ (greedy)
# @   => Match the literal @ symbol
# .*  => Match zero or more characters after @ (greedy)
# \.  => Match a literal dot (.)
# .*  => Match zero or more characters after the dot

# Note: This pattern matches email-like text but does not validate an email.


# Practice 2: Match email-like text ending with .com using $
txt = "akash@gmail.com  bat@gmail.com cat@gmail.com mat@gmail.com dog@yahoo.com"

x = re.findall(r".*@.*\.com$", txt)
print(x)
# Output: ['akash@gmail.com  bat@gmail.com cat@gmail.com mat@gmail.com dog@yahoo.com']

# Breakdown:
# .*     => Match zero or more characters before @ (greedy)
# @      => Match the literal @ symbol
# .*     => Match zero or more characters after @ (greedy)
# \.com  => Match the literal text '.com'
# $      => Match the end of the string (or before a final newline)
# Note: Because .* is greedy and $ anchors the pattern to the end,
#       the entire string is matched as one result.

# Note: This pattern matches email-like text but does not validate an email.
````

## Capture and group


````python
txt = "ababab"
x = re.findall("ab+",txt)
print(x)
````

````python
#email

txt = "abc@gmail.com"

x = re.findall(r"(\w+)@(\w+)\.com",txt)

print(x)

for first,second in x :
  print(first,second)
````

````python
#name ,age

txt = "john 25, Alice 30, Bob 18"

x =re.findall(r"(\w+)\s(\d+)",txt)
print(x)

for name,age in x:
  print(name,age)
````
