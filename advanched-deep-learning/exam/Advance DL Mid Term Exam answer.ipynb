````python
import pandas as pd
import numpy as np
import shutil

import os
import torch
import torchvision
from torchvision import datasets,transforms
import PIL
from PIL import Image
import torch.optim as optim
import torch.nn as nn
from torch.utils.data import DataLoader
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split

````

````python
src ="/content/drive/MyDrive/datasets/faces-age-detection-dataset"
dst = "/content/dataset"
shutil.copytree(src=src,dst=dst)
````

````python

# connect each image ID to its age-group label - part A
## read csv . part-A

df = pd.read_csv("/content/dataset/faces/train.csv")
image_dir = "/content/dataset/faces/Train"

df['image_path'] = df['ID'].apply(
    lambda x :f"{image_dir}/{x}"
)
df['is_exit'] = df["image_path"].apply(
    lambda x : os.path.exists(x)
)

df.head()
````

````python
print(df["is_exit"].value_counts())
````

````python
# show few samples plt - part A
sample_data = df.sample(10)
fig,axes = plt.subplots(2,5,figsize=(12,8))
axes = axes.flatten()

for i in range(10):
  image = Image.open(sample_data['image_path'].iloc[i]).convert("RGB")
  axes[i].imshow(image)
  axes[i].set_title(sample_data['ID'].iloc[i])
  axes[i].axis('off')
plt.tight_layout()
plt.show()
````

````python
# resize , convert to tensor , normalize - part B
IMAGE_RESIZE = 128
BATCH_SIZE = 16
train_transoform = transforms.Compose([
    transforms.Resize((IMAGE_RESIZE,IMAGE_RESIZE)),

    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomRotation(10),

    transforms.ColorJitter(
        brightness=0.2,
        contrast=0.2,
        saturation=0.2
    ),

    transforms.ToTensor(),

    transforms.Normalize(
        mean=[0.5,0.5,0.5],
        std=[0.5,0.5,0.5]
    )
])
eval_transform = transforms.Compose([
    transforms.Resize((IMAGE_RESIZE, IMAGE_RESIZE)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.5, 0.5, 0.5],
        std=[0.5, 0.5, 0.5]
    )
])
````

````python
# encoding - part B
class_label = {
    "YOUNG":0,
    "MIDDLE":1,
    "OLD":2
}
df['label'] = df['Class'].map(class_label)
df.sample(5)
````

````python
#part -c
train_data,val_data = train_test_split(
    df,
    random_state=42,
    test_size=0.02
)

class FaceDataset ():
  def __init__(self,transform,dataframe):
    self.transform = transform
    self.dataframe= dataframe.reset_index(drop=True)
  def __len__(self):
    return len(self.dataframe)
  def __getitem__(self,index):
    row = self.dataframe.iloc[index]
    image = Image.open(row['image_path']).convert('RGB')
    label = int(row['label'])
    if self.transform:
      image = self.transform(image)
    return image,label
train_dataset = FaceDataset(transform=train_transoform,dataframe=train_data)
val_dataset = FaceDataset(transform=eval_transform,dataframe=val_data)



train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=2,
    pin_memory=True
)
val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)
````

Part D: Build the CNN [5 marks]
Build a CNN with four convolutional blocks. Each block should have two convolution layers, then activation and normalization, and a pooling layer at the end. Increase the number of filters across blocks. After the convolutional part, flatten the feature maps and add fully connected layers with dropout. The final layer must have 3 outputs.

````python
images, labels = next(iter(train_loader))

print(images.shape)
print(labels.shape)
````

````python
#part D
class MyCNN(nn.Module):
    def __init__(self, num_class=3):
        super().__init__()

        def block(in_channel, out_channel):
            return nn.Sequential(
                #First convo
                nn.Conv2d(
                    in_channel,
                    out_channel,
                    kernel_size=3,
                    padding=1
                ),
                nn.ReLU(),
                nn.BatchNorm2d(out_channel),

                #Second conv
                nn.Conv2d(
                    out_channel,
                    out_channel,
                    kernel_size=3,
                    padding=1
                ),
                nn.ReLU(),
                nn.BatchNorm2d(out_channel),

                #Pooling
                nn.MaxPool2d(2)
            )

        self.block_1 = block(3, 32)
        self.block_2 = block(32, 64)
        self.block_3 = block(64, 128)
        self.block_4 = block(128, 256)

        self.classifier = nn.Sequential(
            nn.Flatten(),

            nn.Linear(256 * 8 * 8, 128),
            nn.ReLU(),
            nn.Dropout(0.4),

            nn.Linear(128, 64),
            nn.ReLU(),
            nn.Dropout(0.4),

            nn.Linear(64, num_class)
        )

    def forward(self, feature):
        x = self.block_1(feature)
        x = self.block_2(x)
        x = self.block_3(x)
        x = self.block_4(x)

        x = self.classifier(x)

        return x

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Device:", device)
model = MyCNN(num_class=3).to(device)

model
````

Part E: Train and evaluate [4 marks]
Choose a loss function and an optimizer. Train the model, plot or print training/validation loss or accuracy, and briefly explain the role of each part of this pipeline.

````python
#part - E
EPOCHS= 15
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(
    model.parameters(),
    lr=0.001
)

train_accuracies =[]
val_accuracies =[]

val_losses =[]
train_losses = []

for i in range(EPOCHS):

  model.train()

  train_loss_sum = 0.00
  train_correct_sum = 0.00
  train_total_sum = 0.00
  for image ,label in train_loader:
    optimizer.zero_grad()
    image,label = image.to(device),label.to(device)
    output = model(image)
    loss = criterion(output,label)
    loss.backward()
    optimizer.step()
    _, predict = torch.max(output,1)
    train_correct_sum += (predict==label).sum().item()
    train_loss_sum += loss.item()
    train_total_sum += label.size(0)

  train_loss_avg = train_loss_sum/len(train_loader)
  train_accuracy = (train_correct_sum/train_total_sum)*100.00

  train_accuracies.append(train_accuracy)
  train_losses.append(train_loss_avg)

  val_loss_sum = 0.00
  val_correct_sum = 0.00
  val_total_sum = 0.00

  model.eval()
  with torch.no_grad():
    for image,label in val_loader :
      optimizer.zero_grad()
      image,label = image.to(device),label.to(device)
      output = model(image)
      loss = criterion(output,label)
      _,predict = torch.max(output,1)
      val_correct_sum += (predict==label).sum().item()
      val_total_sum+=label.size(0)
      val_loss_sum += loss.item()

  val_loss_avg = val_loss_sum/len(val_loader)
  val_accuracy = (val_correct_sum/val_total_sum)*100.00

  val_accuracies.append(val_accuracy)
  val_losses.append(val_loss_avg)
  print(f"Epoch {i+1}/{EPOCHS} | Train loss :{train_loss_avg}| val loss : {val_loss_avg}")
````

````python
plt.figure(figsize=(8,5))
plt.plot(range(1,EPOCHS+1), train_losses,label="Train Losses")
plt.plot(range(1,EPOCHS+1),val_losses,label="Validation Losses")
plt.ylabel('Loss')
plt.xlabel('Epochs')
plt.title("Loss curve")
plt.legend()
plt.show()
````

````python
plt.figure(figsize=(8,5))
plt.plot(range(1,EPOCHS+1), train_accuracies,label="Train accuracies")
plt.plot(range(1,EPOCHS+1),val_accuracies,label="Validation accuracies")
plt.ylabel('Accuracies')
plt.xlabel('Epochs')
plt.title("Accuracy curve")
plt.legend()
plt.show()
````

Comments and brief : DataLoader loads images in batches. Model performs forward propagation to predict classes. Loss Function (CrossEntropyLoss) measures prediction error. Optimizer (Adam) updates model weights using backpropagation. Training loop repeats forward pass, loss calculation, backward pass, and weight update. Validation loop evaluates model performance on unseen data without updating weights. Accuracy and Loss are recorded to monitor training and detect overfitting.
- in chart we can see both validation and train loss is gradually decrasing
- in same case in validation and train both accuracy is gradually is incrasing which is good.

This pattern can indicate that the model is improving.
