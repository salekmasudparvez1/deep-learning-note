# Numpy Array to tensors

````python
# Numpy arrays → PyTorch tensors (clean & recommended way)
# -------------------------------------------------------
# torch.from_numpy() shares memory with the numpy array.
# .float() converts to torch.float32 (required by most layers).

X_train_tensor = torch.from_numpy(X_train).float()
X_test_tensor  = torch.from_numpy(X_test).float()

y_train_tensor = torch.from_numpy(y_train).float()
y_test_tensor  = torch.from_numpy(y_test).float()

print("X_train_tensor shape:", X_train_tensor.shape)
print("X_test_tensor shape :", X_test_tensor.shape)
print("y_train_tensor shape:", y_train_tensor.shape)
print("y_test_tensor shape :", y_test_tensor.shape)
````

# Defining the Model

````python
X_train.shape[1]
````

````python
import torch.nn as nn
````

````python
class MySimpleNN(nn.Module):
    # MySimpleNN is our own neural network class.
    # nn.Module is the base class provided by PyTorch.
    # By inheriting from nn.Module, PyTorch knows this is a neural network.

    def __init__(self, num_features):
        # __init__ runs when we create the model.
        #
        # num_features = number of input features.
        #
        # Example:
        # If each person has:
        # age, income, height, weight, experience
        #
        # Then:
        # num_features = 5

        super().__init__()
        # Initialize nn.Module.
        #
        # Think of this as:
        # "PyTorch, please initialize all the neural-network
        # functionality for my model."

        self.linear = nn.Linear(num_features, 1)
        # Create a Linear layer.
        #
        # It takes:
        #     num_features inputs
        #
        # and produces:
        #     1 output
        #
        # Mathematically:
        #
        # output = XW + b
        #
        # If num_features = 5:
        #
        #     5 inputs ---> Linear ---> 1 output
        #
        # The Linear layer automatically creates:
        #     weights (W)
        #     bias (b)

        self.sigmoid = nn.Sigmoid()
        # Create a Sigmoid activation function.
        #
        # Sigmoid converts the output into a value between 0 and 1.
        #
        # Example:
        #     3.5  ---> 0.9707
        #    -2.0  ---> 0.1192
        #     0.0  ---> 0.5000
        #
        # This is commonly used for binary classification.

    def forward(self, features):
        # forward() describes how data moves through the network.
        #
        # features = input data.
        #
        # Example:
        # features =
        # [age, income, height, weight, experience]

        out = self.linear(features)
        # Send the input through the Linear layer.
        #
        # features
        #    ↓
        # Linear
        #    ↓
        # out
        #
        # Mathematically:
        # out = features × weights + bias

        out = self.sigmoid(out)
        # Now send the Linear output through Sigmoid.
        #
        # Linear output:
        #     -2.5
        #
        # Sigmoid output:
        #     0.0759
        #
        # So:
        #
        # Linear → Sigmoid

        return out
        # Return the final prediction.
````

# Training Pipeline

                TRAINING
                   │
                   ↓
          ┌─────────────────┐
          │  X_train_tensor │
          └────────┬────────┘
                   │
                   ↓
              ┌─────────┐
              │  MODEL  │
              └────┬────┘
                   │
                   ↓
                y_pred
                   │
                   │ compare
                   ↓
              ┌─────────┐
              │  LOSS   │ ← y_train
              └────┬────┘
                   │
                   ↓
          loss.backward()
                   │
                   ↓
              Gradients
                   │
                   ↓
          optimizer.step()
                   │
                   ↓
            Update weights
                   │
                   ↓
                Repeat

````python
# Create model
model = MySimpleNN(X_train_tensor.shape[1])

# Create optimizer
optimizer = torch.optim.SGD(
    model.parameters(),
    lr=learning_rate
)

# Training loop
for epoch in range(epochs):

    # -------------------------
    # 1. Forward Pass
    # -------------------------
    # Give training features to the model.
    #
    # X_train_tensor:
    # [number_of_samples, number_of_features]
    #
    # Model:
    # Linear → Sigmoid
    #
    # Result:
    # y_pred = predicted probabilities

    y_pred = model(X_train_tensor)


    # -------------------------
    # 2. Calculate Loss
    # -------------------------
    # Compare:
    #
    # y_pred  → model prediction
    # y_train → actual answer
    #
    # Both MUST have the same shape.

    loss = loss_function(
        y_pred,
        y_train_tensor.reshape(-1, 1)
    )


    # -------------------------
    # 3. Zero Gradients
    # -------------------------
    # Remove gradients from the previous iteration.

    optimizer.zero_grad()


    # -------------------------
    # 4. Backward Pass
    # -------------------------
    # Calculate gradients of the loss
    # with respect to model parameters.

    loss.backward()


    # -------------------------
    # 5. Update Parameters
    # -------------------------
    # SGD uses the gradients to update
    # weights and biases.

    optimizer.step()


    # -------------------------
    # 6. Print Loss
    # -------------------------

    print(f"Epoch {epoch + 1}, Loss = {loss.item()}")
````

# Model Evaluation

````python
with torch.no_grad():
  y_pred = model.forward(X_test_tensor)
  y_pred= (y_pred > 0.9).float()
  accuracy = (y_pred == y_test_tensor).float().mean()
  print(f" Accuracy = {accuracy.item()}")
````
