Here is how **Mini-Batch Gradient Descent** fits into the mathematical and practical framework alongside Batch GD and Pure SGD, complete with a comparative mathematical breakdown and code examples for all three in PyTorch.

---

**Mathematical Comparison**

The update rule for parameters $\theta$ using learning rate $\eta$ varies strictly by sample size:

- **Batch GD (Full Dataset):** Uses $N$ total samples.

$$\theta = \theta - \eta \cdot \frac{1}{N} \sum_{i=1}^{N} \nabla_\theta \mathcal{L}(f(x^{(i)}; \theta), y^{(i)})$$

- **Pure Stochastic GD (Single Sample):** Uses 1 sample $i$ chosen at random.

$$\theta = \theta - \eta \cdot \nabla_\theta \mathcal{L}(f(x^{(i)}; \theta), y^{(i)})$$

- **Mini-Batch GD (Subset):** Uses a batch size $B$ (where $1 < B < N$).

$$\theta = \theta - \eta \cdot \frac{1}{B} \sum_{i=1}^{B} \nabla_\theta \mathcal{L}(f(x^{(i)}; \theta), y^{(i)})$$

---

**Detailed Comparison**

| Feature                  | Batch Gradient Descent (GD)            | Pure Stochastic Gradient Descent (SGD) | Mini-Batch Gradient Descent           |
| ------------------------ | -------------------------------------- | -------------------------------------- | ------------------------------------- |
| **Batch Size ($B$)**     | $B = N$ (All samples)                  | $B = 1$                                | $1 < B < N$ (Typically $32, 64, 128$) |
| **Variance of Gradient** | Zero (Exact gradient)                  | Very High (Noisy updates)              | Low to Moderate (Controlled noise)    |
| **Vectorization / GPU**  | High memory overhead                   | Poor GPU hardware utilization          | Optimal GPU parallelization           |
| **Convergence Behavior** | Smooth, guaranteed to local/global min | Trajectory oscillates heavily          | Fast convergence with stable descent  |
| **Use Case**             | Small, in-memory datasets              | Streaming data / Online learning       | Modern Deep Learning standard         |

---

**PyTorch Implementation for All 3 Variants**


```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset

# Setup model and synthetic data
model = nn.Linear(10, 1)
x_data = torch.randn(1000, 10)
y_data = torch.randn(1000, 1)
criterion = nn.MSELoss()
dataset = TensorDataset(x_data, y_data)

# -------------------------------------------------------------
# 1. Batch Gradient Descent (B = N)
# -------------------------------------------------------------
gd_optimizer = optim.SGD(model.parameters(), lr=0.01)

gd_optimizer.zero_grad()
loss = criterion(model(x_data), y_data)
loss.backward()
gd_optimizer.step()

# -------------------------------------------------------------
# 2. Pure Stochastic Gradient Descent (B = 1)
# -------------------------------------------------------------
sgd_loader = DataLoader(dataset, batch_size=1, shuffle=True)
sgd_optimizer = optim.SGD(model.parameters(), lr=0.01)

for batch_x, batch_y in sgd_loader:
    sgd_optimizer.zero_grad()
    loss = criterion(model(batch_x), batch_y)
    loss.backward()
    sgd_optimizer.step()

# -------------------------------------------------------------
# 3. Mini-Batch Gradient Descent (1 < B < N)
# -------------------------------------------------------------
minibatch_loader = DataLoader(dataset, batch_size=64, shuffle=True)
minibatch_optimizer = optim.SGD(model.parameters(), lr=0.01, momentum=0.9)

for batch_x, batch_y in minibatch_loader:
    minibatch_optimizer.zero_grad()
    loss = criterion(model(batch_x), batch_y)
    loss.backward()
    minibatch_optimizer.step()
```
